#Data Dictionary
Este será nuestro diccionario de datos inicial.

Relaciones principales

Relación	Cardinalidad

Vehicles → Trips	1:N

Drivers → Trips	1:N

Vehicles → Telemetry	1:N

Vehicles → Maintenance	1:N

Weather → Trips	N:1 mediante ubicación y tiempo, en el modelo analítico

#Reglas de calidad que definiremos

Antes de procesar los datos, estableceremos algunas reglas que utilizaremos en Silver.

Tabla	| Regla de validación

vehicles |	vehicle_id no puede ser nulo

vehicles |	year debe ser razonable

drivers | 	driver_id no puede estar duplicado

trips |	distance_km debe ser mayor que cero

trips | 	end_time debe ser posterior a start_time

telemetry |	speed_kmh no puede ser negativo

telemetry |	engine_temperature debe estar dentro de rangos definidos

maintenance	repair_cost | no puede ser negativo

weather	precipitation_mm |  no puede ser negativo

#Granularidad de los datos
Esta decisión será especialmente importante para Machine Learning.
Tabla	Granularidad

vehicles |	Un registro por vehículo

drivers |	Un registro por conductor

trips |	Un registro por viaje

telemetry |	Una lectura por vehículo y momento

maintenance |	Un registro por intervención

weather |	Una observación por ubicación y momento

Para nuestro modelo predictivo, posteriormente construiremos variables agregadas por vehículo y ventanas temporales, por ejemplo:
- Temperatura promedio del motor durante las últimas 24 horas.
- RPM máximas durante los últimos siete días.
- Cantidad de códigos de error durante los últimos siete días.
- Horas de funcionamiento desde el último mantenimiento
